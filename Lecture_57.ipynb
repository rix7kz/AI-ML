#np.where()



```python
#np.where(condition,value if true , value if false)
```

```python
import numpy as np
```

```python
marks = np.array([45,72,86,54,32,75,90])
res = np.where(marks>75 , "good","average")
```

```python
print(res)
```

```python
#suppose students below 50 marks get grace of 5 marks
nm = np.where(marks<50 , marks + 5, marks)
```

```python
print(nm)
```

```python
#suppose product costing moree than 1000 recieve extra 10 % discount
prices = np.array([500,1200,800,25001500])
```

```python
discounted = np.where(prices > 1000, prices*0.9, prices )
```

```python
print(discounted)
```
