```python
import numpy as np
```

```python
temperature = np.array([
  28.5, 31.2, 29.8, 35.6, 38.1,
  40.2, 27.9, 30.5, 42.3, 39.8,
  36.4, 33.2, 29.1, 41.7, 44.2,
  31.8, 34.5, 37.6, 26.9, 43.1
])
```

```python
print(temperature)
```

```python
temperature.ndim
```

```python
temperature.shape
```

```python
# Calculate mean, median, minimum and maximum.
avg_t= np.mean(temperature)
print(avg_t)
```

```python
np.min(temperature)
```

```python
np.max(temperature)
```

```python
np.median(temperature)
```

```python
np.std(temperature)
```

```python
np.square(temperature)
```

```python
# Find all temperatures more than one standard deviation above the mean.
new_t = temperature[temperature>(np.mean(temperature)+np.std(temperature))]
print(new_t)
```

```python
np.where(temperature>(np.mean(temperature)+np.std(temperature)),"yes","no")
```

```python
# Calculate the absolute difference between every temperature and the mean.
np.abs(temperature-(np.mean(temperature)))
```

```python
# Find the five readings having the largest deviation from the mean.
largestd=np.sort(np.abs(temperature-(np.mean(temperature))))
print(largestd)
```

```python
largest5=largestd[-5:]
print(largest5)
```

```python
# Round all temperatures to the nearest whole number.
np.round(temperature)
```

```python
cat_t = np.where(temperature >= 40, "Critical", np.where(temperature >= 35, "Warning", "Normal"))
print(cat_t)
```

```python
critical_t = cat_t[ cat_t == "Critical"]
print(critical_t)
```

```python
c=len(critical_t)
print(c)
```

```python
# Reshape the readings into a 4 × 5 array.
reshapet=temperature.reshape(4,5)
print(reshapet)
```

```python
# Find the average temperature for each row.
np.mean(reshapet,axis=1)
```

```python
employees = np.array([
  [101, 42000, 3, 450000],
  [102, 65000, 6, 720000],
  [103, 38000, 2, 300000],
  [104, 90000, 8, 950000],
  [105, 55000, 5, 610000],
  [106, 72000, 7, 830000]
])
```

```python
# Extract salary, experience and annual sales.
salary = employees[:,1]
experience = employees[:,2]
sales = employees[:,3]
```

```python
 #Calculate sales-to-salary ratio.
 sts= np.round(sales/salary)
 sts
```

```python
# Find the employee with the highest annual sales.
employees[np.argmax(sales)]
```

```python
# Find employees having at least 5 years experience AND sales above ₹600,000.
criteria = (experience >= 5) & (sales > 600000)
employees[criteria]
```

```python
bonus = np.where(sales>= 800000, 0.1*sales, np.where(sales >= 600000, 0.07*sales, 0.03*sales))
bonus
```

```python
image = np.array([
  [10, 25, 40, 55, 70, 85],
  [20, 35, 50, 65, 80, 95],
  [30, 45, 60, 75, 90, 105],
  [40, 55, 70, 85, 100, 115],
  [50, 65, 80, 95, 110, 125],
  [60, 75, 90, 105, 120, 135]
])
```

```python
# Extract the central 4 × 4 region.
center= image[1:5, 1:5]
print(center)
```

```python
 #Increase the central region by 20%.
image[1:5,1:5] = image[1:5,1:5]*1.2
image
```

```python
# Cap all pixel values above 100 at 100.
image = np.where(image > 100, 100, image)
image
```

```python
# Find pixels above the 75th percentile.
p75 = np.percentile(image, 75)
p75
```
