```python
prices = np.array([
    499.50,799.90,1299.75,
    2499.50,3999.90,599.25,
    899.80,1599.40
])
```

```python
#round prices to nearest whole number
np.round(prices)
```

```python
#find product counting more than 1000
prices[prices > 1000]
```

```python
#calculate a 15% discount only on price greater than 1000
discounted = np.round((prices[prices > 1000])*0.85)
print(discounted)
```

```python
#calculate the average price of discountd price
np.round(np.mean(discounted))
```

```python
#find the most expensive original product
np.max(prices)
```

```python
np.argmax(prices)
```

```python
#extract the last 4 product
prices[-4:]
```

```python
#calculate the square root of final product and round them to two decimal places
np.round(np.sqrt(discounted),2)
```
