```python
#Problem Statement
#Write a Python program that takes marks of 5 subjects as input.
#Your program should:
#Store the marks in a list.
#Calculate the average marks.
#Find the highest and lowest marks.
#Display the grade using:
```

```python
marks=[]
for i in range(5):
  mark=int(input())
  marks.append(mark)

average=sum(marks)/len(marks)
print("average:",average)
print("max marks:",max(marks))
print("min marks:",min(marks))
```

```python
#word frequency counter
```

```python
s=input()
words = s.split(" ")
print(words)
freq={}
for word in words:
  if word not in freq:
    freq[word]=1
  else:freq[word]+=1
print(freq)
```

```python
#remove duplicate name
s=input()
names=s.split()
print(names)
uniquename=[]
for name in names:
  if name not in uniquename:
    uniquename.append(name)
print(uniquename)
```
