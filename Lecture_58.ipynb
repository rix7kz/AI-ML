```python
import numpy as np
```

```python
employees = np.array([
  [101, 35000, 1200000],
  [102, 48000, 400000],
  [103, 62000, 60000],
  [104, 40000, 130000],
  [105, 75000, 180000],
  [106, 55000, 50000]
])
```

```python
sales = employees[:,2]
```

```python
commission = np.where( sales > 200000 ,"exellent",np.where(sales>150000 , "good", " need improvement"))
```

```python
print(commission)
```

```python
#employees with the sales >= 150000 recieve 5 % commission
icom = np.where( sales>= 150000, sales * 0.05,sales * 0.02)
print(icom)
```

```python
quantity = np.array([2,5,3,8,4,6,10,1,7,5,9,3])
price= np.array([
    120,150,180,200,220,250,270,300,320,350,380,400
])
```

```python
#calculate revenue for every product
rev= quantity*price
print(rev)
```

```python
#calculate the total revenue
np.sum(rev)
```

```python
#find the index of highest revenue maker
np.argmax(rev)
```

```python
#extract the revenue of last 5 product
rev[-4:]
```

```python
avg_rev= np.mean(rev)
```

```python
np.where(rev>avg_rev,"yes","no")
```
