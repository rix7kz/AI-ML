#catching specific errors

```python
try:
  n=int(input())
  print(100/n)

except Exception as e:
  print(e)
```

```python
try:
  n=int(input())
  print(100/n)

except ZeroDivisionError:
  print("division with 0 is not possible")
except ValueError:
  print("please enter numbers only")
except Exception as e:
  print("some error has occured")
```

```python
#else block runs only when no exception occurs
try:
  n=int(input())
  print(100/n)

except ZeroDivisionError:
  print("division with 0 is not possible")
except ValueError:
  print("please enter numbers only")
except Exception as e:
  print("some error has occured")
else :
  print("your number is :",n)
```

```python
#finally block always executes irrespective whether there is an error or not
try:
  n=int(input())
  print(100/n)

except ZeroDivisionError:
  print("division with 0 is not possible")
except ValueError:
  print("please enter numbers only")
except Exception as e:
  print("some error has occured")
else :
  print("your number is :",n)
finally:
  print("program finished")
```

#map

```python
# map(function, iterable) --> it returns a map object --> convert it into list --> list(map(..))
```

```python
numbers = [1, 2, 3, 4]

result = map(lambda x: x * 2, numbers)

print(list(result))
```

```python
# write a code to return a list with the plus 5 of every number in the list.
result = map(lambda x:x+5,numbers)
print(list(result))
```

```python
#Given a list of strings, convert all the strings to all uppercase and display the new list
s=["Aditya","Anshika","Rahul","Raj"]
up=map(lambda s:s.upper(),s)
print(list(up))
```

```python
#Given a list of prices, return a new list with 18% gst included prices.
prices=[100,200,250,400]
# expected: [118, 236, 295, 472]
newprice=map(lambda x:1.18*x,prices)
print(list(newprice))
print(newprice)
```

#filter

```python
#filter() is a built-in Python function that selects elements from an iterable based on a given condition and returns a filter object.

num=[1,2,3,4,5,6,7,8,9,0]
even=list(filter(lambda x:x%2==0,num))
print(even)
```

```python
even=list(filter(lambda x:x%2!=0,num))
print(even)
```

#CSV

```python
import csv

with open("sample.text","w" ,newline="") as file:
  writer= csv.writer(file)
  writer.writerow(["Name","Marks","City"])
  writer.writerow(["Akash","20","Varanasi"])
  writer.writerow(["Garv","19","Varanasi"])
```

```python
#Write a code to read the content from the csv file and display it
with open("sample.text","r") as file:
  reader= csv.reader(file)

  for row in reader:
    print(row)
```
